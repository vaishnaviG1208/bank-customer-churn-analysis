```python
import pandas as pd

df = pd.read_csv('Churn_Modelling.csv')
print(df.head())
print(df.shape)
print(df.columns)
```

```python
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)
print(df.columns)
print(df.shape)
```

```python
print(df.dtypes)
```

```python
from sqlalchemy import create_engine

engine = create_engine('mysql+mysqlconnector://root:Pass@123@127.0.0.1:3306/churn_db')

df.to_sql('customers', con=engine, if_exists='replace', index=False)
print("Data pushed to MySQL successfully!")
```
